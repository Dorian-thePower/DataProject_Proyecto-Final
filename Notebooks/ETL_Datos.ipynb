```python
# Creo un diccionario de nombres masculinos y femeninos para asignar el genero a los nombres desconocidos
male_names = [
    'John','Michael','Charlie','David','James','Robert','William','Richard','Thomas','Charles','Daniel',
    'Matthew','Anthony','Mark','Paul','Steven','Andrew','Joshua','Kevin','Brian','George',
    'Frank','Oscar','Jack','Steve','Bob','Quincy'
]

female_names = [
    'Mary','Patricia','Jennifer','Linda','Elizabeth','Barbara','Susan','Jessica','Sarah','Karen',
    'Nancy','Lisa','Margaret','Betty','Sandra','Ashley','Kimberly','Emily','Donna','Michelle',
    'Grace','Eva','Jane','Alice','Laura','Hannah','Irene','Nina','Rachel'
]
```

```python
# Creo una función para cambiar nombres del dicionaroo a generos 'Hombre' y 'Mujer'.
def gender_from_dict(name):
    first = name.split()[0].capitalize()
    if first in male_names:
        return 'Hombre'
    if first in female_names:
        return 'Mujer'
    return 'Desconocido'
```

```python
# Aplico la función gender_from_dict a los nombres desconocidos para asignarles un genero basado en el diccionario
df_unido.loc[df_unido['Genero'] == 'Desconocido', 'Genero'] = \
    df_unido.loc[df_unido['Genero'] == 'Desconocido', 'Nombre_normalizado'].apply(gender_from_dict)
```

```python
# Sustituyo los valores Desconocido de Zona envio según la Zona envio de la ciudad de la última venta del mismo id_cliente
df_zona_cliente_ref = (
    df_unido.loc[
        df_unido['Zona_envio'].ne('Desconocido'),
        ['Id_cliente', 'Zona_envio', 'Fecha_venta']
    ]
    .assign(
        _fecha_venta=lambda df: pd.to_datetime(
            df['Fecha_venta'],
            dayfirst=True,
            errors='coerce'
        )
    )
    .sort_values('_fecha_venta')
    .drop_duplicates('Id_cliente', keep='last')
    [['Id_cliente', 'Zona_envio']]
)
```

```python
# Reemplazar valores en español
df_unido['Region'] = df_unido['Region'].replace(
    {
        r'(?i)^west$': 'Oeste',
        r'(?i)^south$': 'Sur',
        r'(?i)^north$': 'Norte',
        r'(?i)^east$': 'Este'
    },
    regex=True
)
```

```python
# Reemplazar valores en español
df_unido['Estado'] = (
    df_unido['Estado']
    .astype(str)
    .str.strip()
    .replace({'Shipped': 'Enviado',
              'Completed': 'Completado',
              'Pending': 'Pendiente',
              'Cancelled': 'Cancelado'
              })
)
```

```python
# Normalizo los valores de Nombre. Eliminar espacios, poner letras capitales
df_unido['Nombre'] = (
    df_unido['Nombre']
    .astype(str)
    .str.strip()                
    .str.replace(r'\s+', ' ', regex=True)   
    .str.title()                
)
```

```python
# Voy a añadir otra columna Antiguedad para determinar la duración en años del cliente desde la fecha registro hasta hoy

# Calculo la antigüedad del cliente en años, sin decimales, usando la fecha de registro y la fecha actual
fecha_registro = pd.to_datetime(
    df_unido['Fecha_registro'],
    errors='coerce',
    dayfirst=True
)

df_unido['Antiguedad'] = (
    (pd.Timestamp.today().normalize() - fecha_registro).dt.days // 365.25
)
```
